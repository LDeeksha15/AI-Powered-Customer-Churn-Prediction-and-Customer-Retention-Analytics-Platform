# 6. Feature Engineering — RFM Analysis

---

## Overview

Feature engineering is the process of transforming raw transactional data into meaningful features that can be used for customer analysis and Machine Learning.

For this project, **RFM (Recency, Frequency, Monetary) Analysis** is used to summarize each customer's purchasing behaviour into three key metrics. RFM is one of the most widely adopted customer segmentation techniques in the retail and e-commerce industry.

---

## RFM Metrics

### **Recency (R)**
The number of days since a customer's most recent purchase.

- Lower recency indicates a recently active customer.
- Higher recency may indicate a customer at risk of churning.

---

### **Frequency (F)**
The total number of **unique purchases (Invoices)** made by a customer.

- Customers with higher purchase frequency are generally more loyal.
- Distinct invoice numbers are counted rather than individual line items.

---

### **Monetary (M)**
The total amount of money spent by a customer across all completed purchases.

- Higher monetary value indicates greater contribution to business revenue.
- This metric helps identify high-value customers.

---

## Purpose of RFM Analysis

RFM features are used to:

- Understand customer purchasing behaviour.
- Identify loyal and high-value customers.
- Support customer segmentation.
- Estimate Customer Lifetime Value (CLV).
- Build the customer churn prediction model.

---

## Feature Engineering Process

The cleaned transaction-level dataset is aggregated at the **customer level** to calculate the following features:

- **Recency** → Days since the last purchase.
- **Frequency** → Number of unique invoices.
- **Monetary** → Total spending by the customer.

The resulting **RFM table** serves as the primary input for subsequent customer segmentation, CLV estimation, and Machine Learning model development.

---

```python
# ============================================================
# Step 6.1: Create RFM Features
# ============================================================

# Reference date (One day after the latest transaction)
snapshot_date = df["InvoiceDate"].max() + pd.Timedelta(days=1)

# Create Customer-Level RFM Table
rfm = (
    df.groupby("CustomerID")
      .agg(
          recency=("InvoiceDate",
                   lambda x: (snapshot_date - x.max()).days),

          frequency=("InvoiceNo", "nunique"),

          monetary=("total_amount", "sum"),

          first_purchase=("InvoiceDate", "min"),

          last_purchase=("InvoiceDate", "max"),

          country=("Country", "first"),

          n_unique_products=("StockCode", "nunique")
      )
      .reset_index()
)

print("RFM Table Created Successfully")
print(f"Total Customers : {len(rfm):,}")

rfm.head()
```

```python
# ============================================================
# Step 6.2: Create Additional Customer Features
# ============================================================

# Average Order Value (AOV)
rfm["avg_order_value"] = (
    rfm["monetary"] / rfm["frequency"]
)

# Customer Tenure (Days)
rfm["tenure_days"] = (
    snapshot_date - rfm["first_purchase"]
).dt.days

# Purchase Rate (Orders per Month)
rfm["purchase_rate"] = (
    rfm["frequency"] /
    (rfm["tenure_days"] / 30).clip(lower=1)
)

print("Additional customer features created successfully.")

# Preview new features
rfm[
    [
        "CustomerID",
        "avg_order_value",
        "tenure_days",
        "purchase_rate"
    ]
].head()
```

# 7. Customer Lifetime Value (CLV) Estimation

---

## Overview

Customer Lifetime Value (CLV) is one of the most important business metrics in customer analytics. It estimates the total revenue a customer is expected to generate over their relationship with the business.

Understanding CLV helps organizations identify their most valuable customers and prioritize retention strategies, marketing investments, and personalized customer engagement.

---

## Simplified CLV Formula

For this project, a simplified and easily interpretable CLV formula is used:

\[
\textbf{CLV} = \text{Average Order Value} \times \text{Purchase Rate} \times 12
\]

Where:

- **Average Order Value (AOV):** Average amount spent per purchase.
- **Purchase Rate:** Average number of purchases made per month.
- **12:** Projects the estimated customer value over one year.

---

## Why This Approach?

Although more advanced CLV models such as **BG/NBD**, **Gamma-Gamma**, and **Discounted Cash Flow (DCF)** are widely used in industry, they require additional assumptions and more complex modeling.

For this project, the simplified CLV estimation offers several advantages:

- Easy to understand and interpret.
- Suitable for business stakeholders.
- Simple to implement using transactional data.
- Provides meaningful customer value estimates for segmentation and churn analysis.

---

## Business Value

The estimated Customer Lifetime Value (CLV) will be used to:

- Identify high-value customers.
- Prioritize customer retention efforts.
- Support customer segmentation.
- Recommend personalized marketing strategies.
- Combine with churn prediction to improve business decision-making.

---

```python
# ============================================================
# Step 7.1: Estimate Customer Lifetime Value (CLV)
# ============================================================

# Calculate Customer Lifetime Value (CLV)
# Formula:
# CLV = Average Order Value × Purchase Rate × 12 Months

rfm["clv"] = (
    rfm["avg_order_value"]
    * rfm["purchase_rate"]
    * 12
).round(2)

print("Customer Lifetime Value calculated successfully.\n")

print(f"Average CLV : GBP {rfm['clv'].mean():,.2f}")
print(f"Median CLV  : GBP {rfm['clv'].median():,.2f}")

# Preview CLV values
rfm[
    [
        "CustomerID",
        "avg_order_value",
        "purchase_rate",
        "clv"
    ]
].head()
```

# 8. Customer Segmentation

---

## Overview

Customer segmentation is the process of grouping customers based on their purchasing behaviour. It enables businesses to better understand different customer groups and design targeted marketing and retention strategies.

In this project, customers are segmented using the **RFM (Recency, Frequency, Monetary)** framework.

---

## RFM Scoring

Each customer is assigned a score from **1 to 5** for each RFM metric using **quintile-based ranking**.

- **Recency (R):** Lower recency (more recent purchases) receives a higher score.
- **Frequency (F):** Higher purchase frequency receives a higher score.
- **Monetary (M):** Higher spending receives a higher score.

Each customer therefore receives three scores:

- **R Score (1–5)**
- **F Score (1–5)**
- **M Score (1–5)**

These scores are combined to create an overall **RFM Score**, which represents the customer's overall value and engagement.

---

## Customer Segments

Based on the RFM scores, customers are grouped into business-friendly segments such as:

- Champions
- Loyal Customers
- Potential Loyalists
- At-Risk Customers
- Lost Customers

These segments help businesses identify which customers should be rewarded, nurtured, or targeted with retention campaigns.

---

## Business Value

Customer segmentation enables organizations to:

- Identify high-value customers.
- Detect customers at risk of churn.
- Personalize marketing campaigns.
- Improve customer retention.
- Increase customer lifetime value (CLV).
- Support data-driven business decision-making.

---

```python
# ============================================================
# Step 8: Customer Segmentation using RFM Scores
# ============================================================

# Assign Recency Score (Recent customers receive higher scores)
rfm["R_score"] = pd.qcut(
    rfm["recency"],
    5,
    labels=[5, 4, 3, 2, 1],
    duplicates="drop"
).astype(int)

# Assign Frequency Score (Customers with more purchases receive higher scores)
rfm["F_score"] = pd.qcut(
    rfm["frequency"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

# Assign Monetary Score (Higher spending customers receive higher scores)
rfm["M_score"] = pd.qcut(
    rfm["monetary"],
    5,
    labels=[1, 2, 3, 4, 5],
    duplicates="drop"
).astype(int)

# Calculate Overall RFM Score
rfm["RFM_score"] = (
    rfm["R_score"] +
    rfm["F_score"] +
    rfm["M_score"]
)

# ------------------------------------------------------------
# Create Customer Segments
# ------------------------------------------------------------

def segment(row):

    if row["RFM_score"] >= 13:
        return "Champions"

    elif row["RFM_score"] >= 10:
        return "Loyal Customers"

    elif row["RFM_score"] >= 7:
        return "Potential Loyalists"

    elif row["RFM_score"] >= 5:
        return "At Risk"

    else:
        return "Lost Customers"


# Apply Segmentation
rfm["segment"] = rfm.apply(segment, axis=1)

# Display Segment Distribution
print("Customer Segment Distribution:\n")
print(rfm["segment"].value_counts())

# Preview Results
rfm[
    [
        "CustomerID",
        "R_score",
        "F_score",
        "M_score",
        "RFM_score",
        "segment"
    ]
].head()
```

```python
# ============================================================
# Step 8.1: Customer Segment Distribution
# ============================================================

segment_order = [
    "Champions",
    "Loyal Customers",
    "Potential Loyalists",
    "At Risk",
    "Lost Customers"
]

plt.figure(figsize=(8, 5))

rfm["segment"].value_counts().reindex(segment_order).plot(
    kind="barh",
    color=[
        "#22c55e",
        "#3b82f6",
        "#a78bfa",
        "#f59e0b",
        "#ef4444"
    ]
)

plt.title("Customer Segment Distribution")
plt.xlabel("Number of Customers")
plt.ylabel("Customer Segment")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Step 8.2: Customer Lifetime Value by Segment
# ============================================================

plt.figure(figsize=(10, 5))

sns.boxplot(
    data=rfm,
    x="segment",
    y="clv",
    order=segment_order
)

plt.title("Customer Lifetime Value by Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Customer Lifetime Value (GBP)")

plt.xticks(rotation=30)

# Remove extreme outliers for better readability
plt.ylim(0, rfm["clv"].quantile(0.95))

plt.tight_layout()
plt.show()
```

# 9. Churn Label Definition

---

## Overview

Customer churn refers to customers who have stopped purchasing from the business for a prolonged period. Since the Online Retail dataset does not contain an explicit churn indicator, a churn label must be created using customer purchase history.

For this project, churn is defined based on **Recency**, which measures the number of days since a customer's last purchase.

---

## Churn Definition

A customer is considered **churned** if they have **not placed an order within the last 90 days**.

Mathematically,

- **Recency > 90 days** → Churned (1)
- **Recency ≤ 90 days** → Active (0)

---

## Why 90 Days?

Several recency thresholds were evaluated before selecting the final cutoff.

| Recency Threshold | Approximate Churn Rate |
|------------------:|-----------------------:|
| 60 Days | ~45% |
| **90 Days** | **~33%** |
| 120 Days | ~28% |

A **90-day threshold** provides a balanced class distribution by generating a realistic proportion of churned customers. This helps the Machine Learning model learn meaningful customer behaviour while avoiding excessive class imbalance.

---

## Business Importance

Creating a churn label enables the business to:

- Identify customers at risk of leaving.
- Build predictive Machine Learning models.
- Design proactive customer retention campaigns.
- Improve customer lifetime value (CLV).
- Support data-driven marketing decisions.

---

```python
# ============================================================
# Step 9.1: Create Churn Label
# ============================================================

# Define churn threshold (in days)
CHURN_THRESHOLD_DAYS = 90

# Create churn label
# 1 = Churned
# 0 = Active Customer

rfm["churned"] = (
    rfm["recency"] > CHURN_THRESHOLD_DAYS
).astype(int)

print("Churn labels created successfully.\n")

print(f"Churn Rate : {rfm['churned'].mean() * 100:.2f}%")

# Preview customer churn labels
rfm[
    [
        "CustomerID",
        "recency",
        "churned"
    ]
].head()
```

# 10. Data Leakage Check — A Critical Machine Learning Decision

---

## What is Data Leakage?

Data leakage occurs when information that would not be available at prediction time is used during model training. As a result, the model learns patterns that would not exist in a real-world deployment, producing unrealistically high accuracy and poor generalization.

Preventing data leakage is one of the most important responsibilities of a Machine Learning practitioner.

---

## Potential Leakage in This Project

In this project, the churn label is defined using the following business rule:

- **Recency > 90 days → Churned**
- **Recency ≤ 90 days → Active**

If **Recency** is included as an input feature, the model would simply learn the rule used to create the target variable instead of discovering genuine customer behaviour.

Similarly, **R Score**, which is directly derived from Recency, would also introduce data leakage.

---

## Modeling Decision

To prevent leakage, the following features are **excluded** from model training:

- **Recency**
- **R Score**

Only behaviour-based features that are available before churn occurs are used for prediction.

---

## Why This Matters

Removing leakage ensures that:

- The model learns meaningful customer behaviour instead of memorizing the target.
- Evaluation metrics accurately reflect real-world performance.
- The trained model can be reliably deployed for early churn prediction.
- Business decisions are based on trustworthy predictions.

---

> **Important:** Identifying and preventing data leakage is considered a best practice in Machine Learning and is an important topic frequently discussed during data science and machine learning interviews.

---

```python
excluded_features = ["recency", "R_score"]
print("Features excluded to prevent data leakage:", excluded_features)
```

# 11. Train-Test Split & Data Preprocessing

---

## Overview

After preparing the customer-level dataset and preventing data leakage, the next step is to prepare the data for Machine Learning.

The dataset is divided into **training** and **testing** sets so that the model can be trained on one portion of the data and evaluated on previously unseen customers.

---

## Feature Selection

To avoid data leakage, only customer behaviour features that are available before churn occurs are used for model training.

The following features are selected:

- Frequency
- Monetary
- Average Order Value
- Customer Tenure
- Purchase Rate
- Customer Lifetime Value (CLV)
- Frequency Score
- Monetary Score

The following features are deliberately excluded:

- Recency
- R Score

---

## Data Splitting

The dataset is divided using an **80:20 ratio**:

- **80%** → Training set
- **20%** → Testing set

Stratified sampling is used to preserve the proportion of churned and non-churned customers in both datasets.

---

## Feature Scaling

Machine Learning algorithms often perform better when numerical features are on a similar scale.

Therefore, **StandardScaler** is applied to standardize the selected features before model training.

---

```python
# ============================================================
# Step 11: Train-Test Split & Data Preprocessing
# ============================================================

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# ------------------------------------------------------------
# Select Features and Target Variable
# ------------------------------------------------------------

FEATURES = [
    "frequency",
    "monetary",
    "avg_order_value",
    "tenure_days",
    "purchase_rate",
    "clv",
    "F_score",
    "M_score"
]

X = rfm[FEATURES]
y = rfm["churned"]

# ------------------------------------------------------------
# Split Dataset into Training and Testing Sets
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# ------------------------------------------------------------
# Feature Scaling
# ------------------------------------------------------------

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Data preprocessing completed successfully.\n")

print(f"Training Samples : {X_train.shape[0]:,}")
print(f"Testing Samples  : {X_test.shape[0]:,}")

print(f"\nNumber of Features : {X_train.shape[1]}")
```

# 12. Model Building & Comparison

In this stage, multiple machine learning algorithms are trained and
compared to identify the best-performing model for customer churn
prediction.

Using multiple models helps ensure that the final model is not selected
based on assumptions, but on objective performance evaluation.

### Models Evaluated

- **Logistic Regression**
  - Simple and highly interpretable baseline model.
  - Suitable for understanding the relationship between customer
    behaviour and churn.

- **Random Forest Classifier**
  - An ensemble learning algorithm that combines multiple decision trees.
  - Handles complex and non-linear relationships effectively while
    reducing overfitting.

- **Gradient Boosting Classifier**
  - Sequential ensemble model that focuses on correcting previous errors.
  - Often provides strong predictive performance for structured business
    datasets.

### Evaluation Metrics

Each model is evaluated using multiple performance metrics to obtain a
balanced assessment:

- **Accuracy** – Overall prediction correctness.
- **Precision** – Percentage of predicted churn customers who actually churned.
- **Recall** – Ability to correctly identify actual churn customers.
- **F1-Score** – Harmonic mean of Precision and Recall.
- **ROC-AUC Score** – Measures the model's ability to distinguish
  between churned and non-churned customers.
- **5-Fold Cross Validation** – Evaluates model stability and
  generalization across different subsets of the dataset.

The model with the best combination of predictive performance,
generalization capability, and business interpretability will be selected
for deployment in the Flask web application.

```python
# ============================================================
# Step 12.1: Train Multiple Machine Learning Models
# ============================================================

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier

from sklearn.model_selection import cross_val_score

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    roc_curve
)

# ------------------------------------------------------------
# Define Machine Learning Models
# ------------------------------------------------------------

models = {

    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        C=1.0,
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        max_depth=8,
        random_state=42
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=150,
        random_state=42
    )

}

# ------------------------------------------------------------
# Train and Evaluate Models
# ------------------------------------------------------------

results = []

roc_data = {}

for model_name, model in models.items():

    # Train model
    model.fit(X_train_scaled, y_train)

    # Predictions
    y_pred = model.predict(X_test_scaled)

    # Prediction probabilities
    y_prob = model.predict_proba(X_test_scaled)[:, 1]

    # Cross Validation
    cv_scores = cross_val_score(
        model,
        X_train_scaled,
        y_train,
        cv=5,
        scoring="accuracy"
    )

    # Store evaluation metrics
    results.append({

        "Model": model_name,

        "Accuracy": accuracy_score(y_test, y_pred),

        "Precision": precision_score(y_test, y_pred),

        "Recall": recall_score(y_test, y_pred),

        "F1 Score": f1_score(y_test, y_pred),

        "ROC-AUC": roc_auc_score(y_test, y_prob),

        "Cross Validation Accuracy": cv_scores.mean()

    })

    # Store ROC curve values
    roc_data[model_name] = roc_curve(y_test, y_prob)[:2]

print("All models trained successfully.")
```

```python
# ============================================================
# Step 12.2: Model Performance Comparison
# ============================================================

results_df = (
    pd.DataFrame(results)
    .set_index("Model")
    .round(4)
)

results_df
```

# 13. Model Evaluation

Model evaluation is performed to measure how effectively the trained
models identify customers who are likely to churn.

Since churn prediction is a binary classification problem, relying only
on accuracy can be misleading, especially when the dataset is not
perfectly balanced. Therefore, multiple evaluation metrics are used to
obtain a comprehensive assessment of model performance.

### Evaluation Metrics

- **Accuracy**
  - Measures the overall percentage of correctly classified customers.

- **Precision**
  - Indicates how many customers predicted as churned actually churned.
  - High precision reduces unnecessary retention campaigns.

- **Recall**
  - Measures how many actual churned customers were correctly identified.
  - High recall helps minimize the loss of valuable customers.

- **F1-Score**
  - Balances Precision and Recall into a single metric.
  - Useful when both false positives and false negatives are important.

- **ROC-AUC Score**
  - Evaluates the model's ability to distinguish between churned and
    non-churned customers across different probability thresholds.
  - Values closer to **1.0** indicate better predictive performance.

- **Confusion Matrix**
  - Summarizes the number of correct and incorrect predictions.
  - Helps identify false positives and false negatives.

### Model Selection

The final model is selected based on a combination of predictive
performance, generalization ability, and business interpretability,
rather than relying on a single evaluation metric.

The selected model will be exported and deployed in the Flask web
application to generate real-time customer churn predictions.

```python
# ============================================================
# ROC Curve Comparison
# ============================================================

fig, ax = plt.subplots(figsize=(7, 6))

# Plot ROC curve for each model
for name, (fpr, tpr) in roc_data.items():
    ax.plot(
        fpr,
        tpr,
        linewidth=2,
        label=name
    )

# Random classifier reference line
ax.plot(
    [0, 1],
    [0, 1],
    "k--",
    alpha=0.4,
    label="Random Guess"
)

# Chart formatting
ax.set_title("ROC Curve Comparison", fontsize=14, fontweight="bold")
ax.set_xlabel("False Positive Rate")
ax.set_ylabel("True Positive Rate")

ax.legend(loc="lower right")
ax.grid(True, linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```
